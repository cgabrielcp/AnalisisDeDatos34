# Listas

```python
# Listas de enteros
my_list_1 = [3, 5, 2]
print(my_list_1)


# lista vacía
my_list_2 = []
print(my_list_2)

# Lista Mixta
my_list_3 = [3, "Hello", 44.5]
print(my_list_3)

# Lista Anidada
my_list = ['apple', [3, ["a", 2, 2.5], 8], ['x']]
print(my_list)
```

```python
# insertar elementos a una lista
my_list.insert(1, 30)

my_list
```

```python
my_list_4 = ['x', 'l', 't', 'a']

print(my_list_4[2])
```

```python
# lista Anidada
my_nested_list = ['apple', [3, ["a", 2, 2.5], 8], ['x']]

print(my_nested_list[1][2])
```

```python
# Iterar con una lista
frutas = ['apple', 'banana', 'cherry']

for fruta in frutas:
    print('I like: ', fruta)
```

# Tuplas

```python
# tupla vacia
my_tuple = ()
print(my_tuple)

# tupla int
my_tuple = (3, 2, 5)
print(my_tuple)

# tupla mixta
my_tupla = (3, 2.2, "Texto")
print(my_tupla)

# Tupla anidada
my_tuple = (3, (2, 5), "Texto", [1, 2, 3])
print(my_tuple)
```
